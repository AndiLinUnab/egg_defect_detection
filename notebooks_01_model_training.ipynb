```python
from ultralytics import YOLO

# Cargar modelo base YOLOv8 nano (ligero y rápido para tiempo real)
model = YOLO("yolov8n.pt")

# Entrenar el modelo
results = model.train(
    data="/content/dataset/egg-egg.v34i.yolov8/data.yaml",  # Ruta exacta a tu data.yaml
    epochs=35,  # 35-50 épocas suele dar un excelente Recall/mAP
    imgsz=640,
    batch=16,
    name="egg_detection_v1",
    project="egg_inspection",
)
```
